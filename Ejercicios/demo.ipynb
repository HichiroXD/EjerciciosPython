# Ejercicio Modelo Lineal usando Moore-(algo...)

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
phi = np.array([[1,1], [1,2], [1,3], [1,4], [1,5]])
y = np.array([2, 4, 7, 10, 13]) # Es posible que no haya una ecuación.
phi, y
```

```python
pseudo_inv = np.linalg.inv(phi.T @ phi) @ phi.T
pseudo_inv
```

```python
params = pseudo_inv @ y
np.round(params, 2)
```
