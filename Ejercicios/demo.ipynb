```python
phi = np.array([[1,1], [1,2], [1,3], [1,4], [1,5]])
y = np.array([2, 4, 7, 10, 13]) # Es posible que no haya una ecuación.
phi, y
```

```python
pseudo_inv = np.linalg.inv(phi.T @ phi) @ phi.T
pseudo_inv
```

```python
params = pseudo_inv @ y
np.round(params, 2)
```

```python
# X = (X - X.mean()) / X.std() # Normalización de características
# Y = (Y - Y.mean()) / Y.std() # Normalización de etiquetas

X = np.array([[1], [2], [3]])
```

```python
phi = np.array([np.ones(len(X)), X]).T
phi
```

```python
w = np.zeros((2, 1)) #Pesos iniciales
lr = 0.1 #Tasa de aprendizaje
w, lr
```

```python
epochs = 100 #Número de iteraciones para el entrenamiento
for epoch in range(epochs):
    ols = 0.5 * (y - phi @ w)**2 #Error cuadrático medio
    grad = -phi.T @ (y - phi @ w) #Gradiente
    w = w + lr * grad #Actualización de pesos
w
```
